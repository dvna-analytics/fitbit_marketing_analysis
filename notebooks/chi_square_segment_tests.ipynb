# Chi-Square Segment Tests
We are conducting these tests to validate whether the apparent differences in conversion rate between segments are statistically meaningful.

## Setup
Import packages: `pandas`, `numpy`, `scipy.stats.chi2_contingency`

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
```

## Load and Prepare Data
Load in the `conversion_efficiency_by_platform.csv`, `conversion_by_day_of_week.csv`, and `revenue_by_country` which contain aggregated clicks and conversions by ad platform, day of week, and country.

```python
df_dow = pd.read_csv("../results/conversion_by_day_of_week.csv")
df_dow["non_conversions"] = df_dow["total_clicks"] - df_dow["total_conversions"]
dow_contingency_table = df_dow[["total_conversions","non_conversions"]].values

df_plat = pd.read_csv("../results/conversion_efficiency_by_platform.csv")
df_plat["non_conversions"] = df_plat["total_clicks"] - df_plat["total_conversions"]
plat_contingency_table = df_plat[["total_conversions","non_conversions"]].values

df_country = pd.read_csv("../results/revenue_by_country.csv")
df_country["non_conversions"] = df_country["total_clicks"] - df_country["total_conversions"]
country_contingency_table = df_country[["total_conversions","non_conversions"]].values
```

## Method Overview

### Goal
In the earlier descriptive analysis, some segments appeared to show slightly different conversion rates. This section evaluates whether those differences are statistically meaningful or likely due to sampling variation.

### Outcome Definition
For each segment (ad platform, day of week, and country), we treat each click as resulting in one of two outcomes:  
* **Conversion**
* **Non-conversion**

Using aggregated data, we compute:
* **Conversions** = `total_conversions`
* **Non-conversions** = `total_clicks` - `total_conversions`

### Statistical Test
We use a **Chi-square test of independence** to test whether the conversion outcome is dependent on the categorical variable (e.g., country).  

**Null hypothesis (H<sub>0</sub>):** Conversion outcome is independent of the categorical variable (i.e. conversion rates do not differ by group).  
**Alternative hypothesis (H<sub>1</sub>):** Conversion outcome depends on the categorical variable (i.e. conversion rates differ for at least one group).

### Significance Threshold
We use a significance level of **&alpha; = 0.05**.
* If **p < 0.05**, we reject H<sub>0</sub> and conclude that conversions are associated with the categorical variable.
* If **p >= 0.05**, we fail to reject H<sub>0</sub> and conclude there is insufficient evidence of dependence in this dataset.

### Assumption Check
A standard assumption of the chi-square test is that **expected cell counts are sufficiently large**. For each test, we verify that all expected frequencies returned by `chi2_contingency` are greater than 5.

### Notes on Interpretation
Failing to reject H<sub>0</sub> does not prove conversion rates are identical across groups. It means that within this sample and aggregation level, any observed differences are not statistically distinguishable from random variation at the chosen significance threshold.

## Segment-Level Statistical Testing

| **Segment** | **&chi;<sup>2</sup>** | **df** | **p-value** | **Conclusion**
|---|---|---|---|---|
|Ad Platform| 0.3307 | 3 | 0.9542 | No dependence
|Country| 2.7295 | 5 | 0.7416 | No dependence
|Day of Week| 2.8671 | 6 | 0.8253 | No dependence

Across all tested segments, conversion outcome was not statistically associated with the categorical group at &alpha; = 0.05. Observed performance differences are likely attributable to sampling variation within this dataset.

```python
## Chi-square test for Day of Week
chi2, p_value, dof, expected = chi2_contingency(dow_contingency_table)
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.4f}")
print(f"Minium expected: {expected.min():.1f}")
```

```python
## Chi-square test for Country
chi2, p_value, dof, expected = chi2_contingency(country_contingency_table)
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.4f}")
print(f"Minium expected: {expected.min():.1f}")
```

```python
## Chi-square test for Ad Platform
chi2, p_value, dof, expected = chi2_contingency(plat_contingency_table)
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.4f}")
print(f"Minium expected: {expected.min():.1f}")
```
